### **Required Libraries**

```python
import pandas as pd
import numpy as np
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from sklearn.model_selection import train_test_split

nltk.download('stopwords')
nltk.download('wordnet')
```

### **Data Gathering**

```python
df = pd.read_csv("/content/WELFake_Dataset.csv.zip")
df.head()
```

```python
df = df.dropna() #handle missing value by drop function
```

```python
df.reset_index(inplace=True)
df.head()
```

```python
df = df.drop(['text'],axis = 1)
```

### **Data Preprocessing**

```python
lm = WordNetLemmatizer()
corpus = []
for i in range(0,len(df)):
  review = re.sub('[^a-zA-Z0-9]',' ', df['title'][i])
  review = review.lower()  #lowercase
  review = review.split()  #Tokenization
  review = [lm.lemmatize(x) for x in review if x not in stopwords.words('english')] #remove stopwards and lemmatization
  review = ' '.join(review)  #join syntax
  corpus.append(review)
```

### **Vectorization**

```python
tf = TfidfVectorizer()
x = tf.fit_transform(corpus)
x
```

```python
y= df['label']
y.head()
```

### **Data splitting into the train and test**

```python

x_train, x_test,y_train, y_test = train_test_split(x,y, test_size = 0.3,random_state = 10,stratify = y)
```

```python
x_train.shape[0], len(y_train)
```

```python
x_test.shape[0], len(y_test)
```

### **Model Building**

```python
rf = RandomForestClassifier()
rf.fit(x_train,y_train)
```

### **Model Evaluation**

```python
y_pred = rf.predict(x_test)
accuracy_score = accuracy_score(y_test,y_pred)
accuracy_score
```

```python
class Evaluation:
  def __init__(self,model,x_train,x_test,y_train,y_test):
    self.model = model
    self.x_train = x_train
    self.x_test = x_test
    self.y_train = y_train
    self.y_test = y_test

  def train_evaluation(self):
    y_pred_train = self.model.predict(self.x_train)
    acc_scr_train = accuracy_score(self.y_train,y_pred_train)
    print("Accuracy Score On Training Data Set :",acc_scr_train)
    print()

    con_mat_train = confusion_matrix(self.y_train,y_pred_train)
    print("Confusion Matrix On Training Data Set :\n",con_mat_train)
    print()

    class_rep_train = classification_report(self.y_train,y_pred_train)
    print("Classification On Training Data Set :\n",class_rep_train)
    print()

  def test_evaluation(self):
    y_pred_test = self.model.predict(self.x_test)
    acc_scr_train = accuracy_score(self.y_test,y_pred_test)
    print("Accuracy Score On Testing Data Set :",acc_scr_train)
    print()

    con_mat_train = confusion_matrix(self.y_test,y_pred_test)
    print("Confusion Matrix On Testing Data Set :\n",con_mat_train)
    print()

    class_rep_train = classification_report(self.y_test,y_pred_test)
    print("Classification On Testing Data Set :\n",class_rep_train)
    print()
```

```python
#Checking the accuracy on training dataset
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
Evaluation(rf,x_train, x_test, y_train, y_test).train_evaluation()
```

```python
#Checking the accuracy on testing dataset
Evaluation(rf,x_train, x_test, y_train, y_test).test_evaluation()
```

### **Prediction Pipeline**

```python
class Preprocessing:
  def __init__(self,data):
    self.data = data

  def text_preprocessing_user(self):
    lm = WordNetLemmatizer()
    pred_data = [self.data]
    preprocess_data = []
    for data in pred_data:
      review = re.sub('[^a-zA-Z0-9]',' ', data)
      review = review.lower()  #lowercase
      review = review.split()  #Tokenization
      review = [lm.lemmatize(x) for x in review if x not in stopwords.words('english')] #remove stopwards and lemmatization
      review = ' '.join(review)  #join syntax
      preprocess_data.append(review)
    return preprocess_data
```

```python
data = 'LAW ENFORCEMENT ON HIGH ALERT Following Threats Against Cops And Whites On 9-11By #BlackLivesMatter And #FYF911 Terrorists [VIDEO]'
Preprocessing(data).text_preprocessing_user()
```

```python
class Prediction:
  def __init__(self,pred_data,model):
    self.pred_data = pred_data
    self.model = model

  def prediction_model(self):
    preprocess_data = Preprocessing(self.pred_data).text_preprocessing_user()
    data = tf.transform(preprocess_data)
    prediction = self.model.predict(data)

    if prediction [0] == 0:
      return "The News Is Fake"
    else:
      return "The News Is Real"
```

```python
data = 'LAW ENFORCEMENT ON HIGH ALERT Following Threats Against Cops And Whites On 9-11By #BlackLivesMatter And #FYF911 Terrorists [VIDEO]'
Prediction(data,rf).prediction_model()
```
