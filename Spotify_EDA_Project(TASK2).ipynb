### **Data Cleaning**

```python
#Handle missing values appropriately
for col in df.columns:
    missing_val = df[col].isnull().mean()  # % of missing values

    if df[col].dtype == "object":  # Categorical
        if missing_val < 0.05:
            df[col] = df[col].fillna(df[col].mode()[0])  # Fill with mode
        else:
            df[col] = df[col].fillna("Unknown")  # Fill with 'Unknown
    else:  # Numeric
        if missing_val < 0.05:
            df[col] = df[col].fillna(df[col].mean())  # Fill with mean
        else:
            df[col] = df[col].fillna(df[col].median())  # Fill with median
```

### **Data Visualization**

```python
#Histogram
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df["popularity"], color='orange')
plt.title("Histogram of popularity")
plt.show()
```

```python
#Boxplot
sns.boxplot(df["liveness"], color='yellow')
plt.title("Boxplot of liveness")
plt.show()
```

```python
#Correlation Heatmap
corr = df.select_dtypes(include='number').corr()

plt.figure(figsize=(7,5))
sns.heatmap(corr, annot= True ,cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
#cmap="coolwarm" → scale (blue = positive correlation, red = negative correlation)।
```

```python
#BAR PLOT
genre_count = df["genre"].value_counts()

plt.figure(figsize=(9,4))
sns.barplot(x=genre_count.index, y=genre_count.values)
plt.title("Most Common genre")
plt.xlabel("genre")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
artist_name_count = df["artist_name"].value_counts().head(10)

plt.figure(figsize=(6,4))
sns.barplot(x=artist_name_count.index, y=artist_name_count.values)
plt.title("10 TOP artists")
plt.xlabel("Artist")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="energy", y="danceability", data =df,color="purple")
plt.title("Scatter Plot: Energy vs Danceability")
plt.xlabel("Energy")
plt.ylabel("Danceability")
plt.show()
```
